## Broadcast in Operations

What we did before,

```python
x = torch.FloatTensor([[1, 2]])  # 텐서 생성
y = torch.FloatTensor([[4, 8]])  # 텐서 생성

print(x.size())  # 결과 출력
print(y.size())  # 결과 출력
```

```python
z = x + y  # 변수 할당
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

Broadcast feature provides operations between different shape of tensors.

### Tensor + Scalar

```python
x = torch.FloatTensor([[1, 2],  # 텐서 생성
                       [3, 4]])  # 코드 실행
y = 1  # 변수 할당

print(x.size())  # 결과 출력
```

```python
z = x + y  # 변수 할당
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

### Tensor + Vector

```python
x = torch.FloatTensor([[1, 2],  # 텐서 생성
                       [4, 8]])  # 코드 실행
y = torch.FloatTensor([3,  # 텐서 생성
                       5])  # 코드 실행

print(x.size())  # 결과 출력
print(y.size())  # 결과 출력
```

```python
z = x + y  # 변수 할당
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

```python
x = torch.FloatTensor([[[1, 2]]])  # 텐서 생성
y = torch.FloatTensor([3,  # 텐서 생성
                       5])  # 코드 실행

print(x.size())  # 결과 출력
print(y.size())  # 결과 출력
```

```python
z = x + y  # 변수 할당
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

### Tensor + Tensor

```python
x = torch.FloatTensor([[1, 2]])  # 텐서 생성
y = torch.FloatTensor([[3],  # 텐서 생성
                       [5]])  # 코드 실행

print(x.size())  # 결과 출력
print(y.size())  # 결과 출력
```

```python
z = x + y  # 변수 할당
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

## Tensor Shaping

### reshape: Change Tensor Shape

```python
x = torch.FloatTensor([[[1, 2],  # 텐서 생성
                        [3, 4]],  # 코드 실행
                       [[5, 6],  # 코드 실행
                        [7, 8]],  # 코드 실행
                       [[9, 10],  # 코드 실행
                        [11, 12]]])  # 코드 실행

print(x.size())  # 결과 출력
```

```python
print(x.reshape(12)) # 12 = 3 * 2 * 2  # 결과 출력
print(x.reshape(-1))  # 결과 출력
```

```python
print(x.reshape(3, 4)) # 3 * 4 = 3 * 2 * 2  # 결과 출력
print(x.reshape(3, -1))  # 결과 출력
```

```python
print(x.reshape(3, 1, 4))  # 결과 출력
print(x.reshape(-1, 1, 4))  # 결과 출력
```

```python
print(x.reshape(3, 2, 2, 1))  # 결과 출력
```

### squeeze: Remove dimension which has only one element.

```python
x = torch.FloatTensor([[[1, 2],  # 텐서 생성
                        [3, 4]]])  # 코드 실행
print(x.size())  # 결과 출력
```

Remove any dimension which has only one element.

```python
print(x.squeeze())  # 결과 출력
print(x.squeeze().size())  # 결과 출력
```

Remove certain dimension, if it has only one element.
If it is not, there would be no change.

```python
print(x.squeeze(0).size())  # 결과 출력
print(x.squeeze(1).size())  # 결과 출력
```

### unsqueeze: Insert dimension at certain index.

```python
x = torch.FloatTensor([[1, 2],  # 텐서 생성
                       [3, 4]])  # 코드 실행
print(x.size())  # 결과 출력
```

```python
print(x.unsqueeze(2))  # 결과 출력
print(x.unsqueeze(-1))  # 결과 출력
print(x.reshape(2, 2, -1))  # 결과 출력
```

## Slicing and Concatenation

### Indexing and Slicing

Prepare target tensor.

```python
x = torch.FloatTensor([[[1, 2],  # 텐서 생성
                        [3, 4]],  # 코드 실행
                       [[5, 6],  # 코드 실행
                        [7, 8]],  # 코드 실행
                       [[9, 10],  # 코드 실행
                        [11, 12]]])  # 코드 실행
print(x.size())  # 결과 출력
```

Access to certain dimension.

```python
print(x[0])  # 결과 출력
print(x[0, :])  # 결과 출력
print(x[0, :, :])  # 결과 출력
```

```python
print(x[-1])  # 결과 출력
print(x[-1, :])  # 결과 출력
print(x[-1, :, :])  # 결과 출력
```

```python
print(x[:, 0, :])  # 결과 출력
```

Access by range. Note that the number of dimensions would not be changed.

```python
print(x[1:3, :, :].size())  # 결과 출력
print(x[:, :1, :].size())  # 결과 출력
print(x[:, :-1, :].size())  # 결과 출력
```

### split: Split tensor to desirable shapes.

```python
x = torch.FloatTensor(10, 4)  # 텐서 생성
```

```python
splits = x.split(4, dim=0)  # 텐서 분할

for s in splits:  # 텐서 분할
    print(s.size())  # 결과 출력
```

### chunk: Split tensor to number of chunks.

```python
x = torch.FloatTensor(8, 4)  # 텐서 생성
```

```python
chunks = x.chunk(3, dim=0)  # 텐서 조각내기

for c in chunks:  # 텐서 조각내기
    print(c.size())  # 결과 출력
```

### index_select: Select elements by using dimension index.

```python
x = torch.FloatTensor([[[1, 1],  # 텐서 생성
                        [2, 2]],  # 코드 실행
                       [[3, 3],  # 코드 실행
                        [4, 4]],  # 코드 실행
                       [[5, 5],  # 코드 실행
                        [6, 6]]])  # 코드 실행
indice = torch.LongTensor([2, 1])  # 텐서 생성

print(x.size())  # 결과 출력
```

```python
y = x.index_select(dim=0, index=indice)  # 인덱스 선택

print(y)  # 결과 출력
print(y.size())  # 결과 출력
```

### cat: Concatenation of multiple tensors in the list.

```python
x = torch.FloatTensor([[1, 2, 3],  # 텐서 생성
                       [4, 5, 6],  # 코드 실행
                       [7, 8, 9]])  # 코드 실행
y = torch.FloatTensor([[10, 11, 12],  # 텐서 생성
                       [13, 14, 15],  # 코드 실행
                       [16, 17, 18]])  # 코드 실행

print(x.size(), y.size())  # 결과 출력
```

```python
z = torch.cat([x, y], dim=0)  # 텐서 결합
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

```python
z = torch.cat([x, y], dim=-1)  # 텐서 결합
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

### stack: Stacking of multiple tensors in the list.

```python
z = torch.stack([x, y])  # 텐서 쌓기
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

Or you can specify the dimension. Default is 0.

```python
z = torch.stack([x, y], dim=-1)  # 텐서 쌓기
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

### Implement 'stack' function by using 'cat'.

```python
# z = torch.stack([x, y])
z = torch.cat([x.unsqueeze(0), y.unsqueeze(0)], dim=0)  # 차원 추가
print(z)  # 결과 출력
print(z.size())  # 결과 출력
```

### Useful Trick: Merge results from iterations

```python
result = []  # 변수 할당
for i in range(5):  # 코드 실행
    x = torch.FloatTensor(2, 2)  # 텐서 생성
    result += [x]  # 변수 할당

result = torch.stack(result)  # 변수 할당
result.size()  # 텐서 크기 확인
```

### expand: copy the given tensor and concat those at desired dimension.

```python
x = torch.FloatTensor([[[1, 2]],  # 텐서 생성
                       [[3, 4]]])  # 코드 실행
print(x.size())  # 결과 출력
```

```python
y = x.expand(*[2, 3, 2])  # 텐서 확장

print(y)  # 결과 출력
print(y.size())  # 결과 출력
```

#### Implement expand with cat.

```python
y = torch.cat([x, x, x], dim=1)  # 텐서 결합

print(y)  # 결과 출력
print(y.size())  # 결과 출력
```

### randperm: Random Permutation

```python
x = torch.randperm(10)  # 랜덤 순열 생성

print(x)  # 결과 출력
print(x.size())  # 결과 출력
```

### argmax: Return index of maximum values

```python
x = torch.randperm(3**3).reshape(3, 3, -1)  # 랜덤 순열 생성

print(x)  # 결과 출력
print(x.size())  # 결과 출력
```

```python
y = x.argmax(dim=-1)  # 최대값 인덱스

print(y)  # 결과 출력
print(y.size())  # 결과 출력
```

### topk: Return tuple of top-k values and indices.

```python
values, indices = torch.topk(x, k=1, dim=-1)  # 상위 K개 추출

print(values.size())  # 결과 출력
print(indices.size())  # 결과 출력
```

Note that topk didn't reduce the dimension, even in $k=1$ case.

```python
print(values.squeeze(-1))  # 결과 출력
print(indices.squeeze(-1))  # 결과 출력
```

```python
print(x.argmax(dim=-1) == indices.squeeze(-1))  # 결과 출력
```

```python
_, indices = torch.topk(x, k=2, dim=-1)  # 상위 K개 추출
print(indices.size())  # 결과 출력

print(x.argmax(dim=-1) == indices[:, :, 0])  # 결과 출력
```

### Sort by using topk

```python
target_dim = -1  # 변수 할당
values, indices = torch.topk(x,  # 상위 K개 추출
                             k=x.size(target_dim),  # 변수 할당
                             largest=True)  # 변수 할당

print(values)  # 결과 출력
```

### Topk by using sort

```python
k=1  # 변수 할당
values, indices = torch.sort(x, dim=-1, descending=True)  # 텐서 정렬
values, indices = values[:, :, :k], indices[:, :, :k]  # 변수 할당

print(values.squeeze(-1))  # 결과 출력
print(indices.squeeze(-1))  # 결과 출력
```

### masked_fill: fill the value if element of mask is True.

```python
x = torch.FloatTensor([i for i in range(3**2)]).reshape(3, -1)  # 텐서 생성

print(x)  # 결과 출력
print(x.size())  # 결과 출력
```

```python
mask = x > 4  # 변수 할당

print(mask)  # 결과 출력
```

```python
y = x.masked_fill(mask, value=-1)  # 마스크로 채우기

print(y)  # 결과 출력
```
