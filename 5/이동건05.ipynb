```python
import torch # 파이토치 임포트
x = torch.FloatTensor([[1, 2], [3, 4]]).requires_grad_(True) # 미분 가능 텐서 생성
x1 = x + 2 # 덧셈 연산
x2 = x - 2 # 뺄셈 연산
x3 = x1 * x2 # 곱셈 연산
y = x3.sum() # 합계 연산
y.backward() # 역전파 수행
print(x.grad) # 기울기 출력
```

```python
import torch.nn.functional as F # 함수형 API 임포트
target = torch.FloatTensor([[.1, .2, .3], [.4, .5, .6], [.7, .8, .9]]) # 목표 텐서 설정
x = torch.rand_like(target, requires_grad=True) # 랜덤 초기값 설정
learning_rate = 1. # 학습률 설정
threshold = 1e-5 # 임계값 설정
while True: # 무한 루프 시작
    loss = F.mse_loss(x, target) # 손실 계산
    loss.backward() # 기울기 계산
    with torch.no_grad(): # 기울기 업데이트 모드
        x -= learning_rate * x.grad # 가중치 업데이트
        x.grad.zero_() # 기울기 초기화
    if loss < threshold: # 종료 조건 확인
        break # 루프 종료
print(x) # 최종값 출력
```

```python
import pandas as pd # 판다스 임포트
import torch.nn as nn # 신경망 모듈 임포트
import torch.optim as optim # 최적화 모듈 임포트
from sklearn.datasets import fetch_openml # 데이터셋 로드 함수 임포트
boston = fetch_openml(name='boston', version=1, as_frame=True) # 보스턴 데이터 로드
df = pd.DataFrame(boston.data, columns=boston.feature_names) # 데이터프레임 생성
df["TARGET"] = boston.target # 타겟 컬럼 추가
cols = ["TARGET", "INDUS", "RM", "LSTAT", "NOX", "DIS"] # 특성 선택
data = torch.from_numpy(df[cols].values).float() # 텐서 변환
y = data[:, :1] # 레이블 추출
x = data[:, 1:] # 특성 추출
n_epochs = 2000 # 에포크 설정
learning_rate = 1e-3 # 학습률 설정
model = nn.Linear(x.size(-1), y.size(-1)) # 선형 모델 정의
optimizer = optim.SGD(model.parameters(), lr=learning_rate) # SGD 설정
for i in range(n_epochs): # 학습 루프
    y_hat = model(x) # 예측값 계산
    loss = F.mse_loss(y_hat, y) # 손실 측정
    optimizer.zero_grad() # 기울기 초기화
    loss.backward() # 역전파 수행
    optimizer.step() # 가중치 업데이트
    if (i + 1) % 500 == 0: # 중간 출력 조건
        print('Epoch %d: loss=%.4e' % (i + 1, loss)) # 손실 출력
print("최종 결과 예측 완료") # 완료 메시지
```
