## 1. 환경 설정

```python
import torch # 파이토치 기본 라이브러리 임포트
import torch.nn as nn # 신경망 구축을 위한 모듈 임포트
import torch.nn.functional as F # 자주 사용되는 신경망 함수 임포트
```

## 2. 4장 선형 계층 - 직접 구현

```python
W = torch.FloatTensor([[1, 2], [3, 4], [5, 6]]) # 3x2 크기의 가중치 행렬 정의
b = torch.FloatTensor([2, 2]) # 크기가 2인 편향 벡터 정의
x = torch.FloatTensor([[1, 1, 1], [2, 2, 2], [3, 3, 3], [4, 4, 4]]) # 4x3 크기의 입력 데이터 정의

def linear(x, W, b): # 선형 연산 함수 정의
    y = torch.matmul(x, W) + b # 행렬 곱 연산 후 편향 합산 수행
    return y # 계산된 결과 반환

y_manual = linear(x, W, b) # 정의한 함수로 출력값 계산
print(y_manual.size()) # 출력 데이터의 크기 확인
```

## 3. 4장 선형 계층 - nn.Module 구현

```python
class MyLinear(nn.Module): # nn.Module을 상속받는 사용자 정의 클래스 선언
    def __init__(self, input_dim=3, output_dim=2): # 생성자를 통한 초기화 과정 정의
        super().__init__() # 부모 클래스인 nn.Module의 초기화 수행
        self.W = nn.Parameter(torch.FloatTensor(input_dim, output_dim)) # 학습 가능한 가중치 파라미터 등록
        self.b = nn.Parameter(torch.FloatTensor(output_dim)) # 학습 가능한 편향 파라미터 등록

    def forward(self, x): # 순전파 연산 과정 정의
        y = torch.matmul(x, self.W) + self.b # 입력 데이터와 가중치 곱 연산 및 편향 합산
        return y # 최종 연산 결과 반환

linear_model = MyLinear(3, 2) # 모델 객체 생성 및 입출력 차원 설정
y_custom = linear_model(x) # 모델을 통한 예측값 계산 수행
print(y_custom.size()) # 결과 텐서의 크기 출력
```

## 4. 4장 선형 계층 - nn.Linear 사용

```python
builtin_linear = nn.Linear(3, 2) # 파이토치 내장 선형 계층 모듈 생성
y_builtin = builtin_linear(x) # 내장 모듈을 이용한 연산 결과 도출
print(y_builtin.size()) # 결과 데이터의 규격 확인
```

## 5. 5장 손실 함수 - MSE 구현 및 사용

```python
def mse(x_hat, x): # 평균 제곱 오차 계산 함수 정의
    y = ((x - x_hat)**2).mean() # 오차 제곱의 평균 계산 공식 적용
    return y # 계산된 손실값 반환

target = torch.FloatTensor([[1, 1], [2, 2]]) # 정답 데이터를 담은 텐서 정의
prediction = torch.FloatTensor([[0, 0], [0, 0]]) # 예측 데이터를 담은 텐서 정의

print(mse(prediction, target)) # 직접 구현한 함수로 손실값 측정
print(F.mse_loss(prediction, target)) # 함수형 API를 이용한 손실값 측정

mse_loss = nn.MSELoss() # 손실 함수 객체 생성
print(mse_loss(prediction, target)) # 생성된 객체를 통한 최종 손실값 측정
```
